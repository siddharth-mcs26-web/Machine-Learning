```python
import numpy as np
import matplotlib.pyplot as plt

X = np.random.randint(1, 10, 1000)
true_w = 5
true_b = 89
Y = X*true_w + true_b
w = 0
b = 0
n = len(X)
epochs = 1000
learning_rate = 0.01
loss_train = []
for i in range(epochs):
    y_pred = w * X + b
    error = y_pred - Y
    loss = np.mean(error**2)
    loss_train.append(loss)
    dw = 2/n * np.sum(X*error)
    db = 2/n * np.sum(error)
    
    w = w - learning_rate*dw
    b = b - learning_rate*db

x_axis = [i for i in range(epochs)]
plt.plot(x_axis, loss_train)
plt.show()
print(w, b)
```
