```python
import numpy as np
X = np.array([1,2,3,4,5])
Y = np.array([3,5,7,9,11])

w = 0
b = 0
n = len(X)
epochs = 1000
learning_rate = 0.01
for i in range(epochs):
    y_pred = w * X + b
    error = y_pred - Y
    loss = np.mean(error**2)
    
    dw = 2/n * np.sum(X*error)
    db = 2/n * np.sum(error)
    
    w = w - learning_rate*dw
    b = b - learning_rate*db
    
    print(np.mean(error))
```
