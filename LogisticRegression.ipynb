```python
import numpy as np

X = np.random.randint(1, 100, 1000)
Y = []
for i in X:
    if i > 80:
        Y.append(1)
    else:
        Y.append(0)
Y = np.array(Y)
X_train = X[:800]
Y_train = Y[:800]
X_test = X[800:]
Y_test = Y[800:]

n = 800
def sigmoid(Z):
    return 1/(1+np.exp(-Z))

w = 0
b = 0
epochs = 1000
alpha = 0.01
for i in range(epochs):
    Z = X_train*w + b
    y_pred = np.vectorize(sigmoid)(Z)
    error = y_pred - Y_train
    dw = 1/n*np.sum(error*X_train)
    db = 1/n*np.sum(error)

    w = w - alpha*dw
    b = b - alpha*db

Z = X_test*w + b
y_pred = np.vectorize(sigmoid)(Z)
predictions = []
for y in y_pred:
    if y > 0.5:
        predictions.append(1)
    else:
        predictions.append(0)

accuracy = []
for i in range(len(predictions)):
    if (predictions[i]==Y_test[i]):
        accuracy.append(1)
    else:
        accuracy.append(0)

acc = np.sum(accuracy)/200
print(acc)











```
