```python
import numpy as np
import matplotlib.pyplot as plt
import random
random.seed(42)
X = np.random.randint(1, 10000, 100)
Y = []
for i in X:
    if i > 3300:
        Y.append(1)
    else:
        Y.append(0)
Y = np.array(Y)
split_boundary = int(len(X)*(0.8))

X_train = X[:split_boundary]
Y_train = Y[:split_boundary]
X_test = X[split_boundary:]
Y_test = Y[split_boundary:]

def sigmoid(Z):
    return 1/(1+np.exp(-Z))

def loss(y, y_pred):
    epsilon = 1e-15
    
    y_pred = np.clip(y_pred, epsilon, 1 - epsilon)
    
    return -np.mean(
        y * np.log(y_pred) +
        (1 - y) * np.log(1 - y_pred)
    )

w = 0
b = 0
epochs = 1000
alpha = 0.00000001
loss_train = []
for i in range(epochs):
    Z = X_train*w + b
    y_pred = np.vectorize(sigmoid)(Z)
    error = y_pred - Y_train
    loss_train.append(loss(Y_train, y_pred))
    dw = np.mean(np.sum(error*X_train))
    db = np.mean(np.sum(error))

    w = w - alpha*dw
    b = b - alpha*db

Z = X_test*w + b
y_pred = np.vectorize(sigmoid)(Z)
predictions = []
for y in y_pred:
    if y > 0.5:
        predictions.append(1)
    else:
        predictions.append(0)

accuracy = []
for i in range(len(predictions)):
    if (predictions[i]==Y_test[i]):
        accuracy.append(1)
    else:
        accuracy.append(0)

acc = (np.sum(accuracy))/len(accuracy)
print(acc)
x_axis = [i for i in range(epochs)]
plt.plot(x_axis, loss_train)
plt.show()
```
