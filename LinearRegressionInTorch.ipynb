```python
import torch
import random
import matplotlib.pyplot as plt
random.seed(42)

x = torch.rand((10))
true_w = 3
true_b = 4
y = x*true_w + true_b
w = torch.tensor(0. , requires_grad=True)
b = torch.tensor(0. , requires_grad=True)
learning_rate = 0.01
epochs = 10
loss_plot = []
for i in range(epochs):
    prediction_y = x*w + b
    loss = ((prediction_y - y)**2).mean()
    loss_plot.append(loss.item())
    loss.backward()

    with torch.no_grad():
        w -= learning_rate*w.grad
        b -= b - learning_rate*b.grad
    
    #loss_plot.append(loss.item())
    w.grad = None
    b.grad = None
    
plt.plot(loss_plot)
plt.show()
print(w)
print(b)
    
```
